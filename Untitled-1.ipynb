```python
# Temperatura

temperatura_promedio = df["Temperatura (°C)"].mean()
temperatura_maxima = df["Temperatura (°C)"].max()
temperatura_minima = df["Temperatura (°C)"].min()

# Humedad ambiental

humedad_promedio = df["Humedad (%)"].mean()
humedad_maxima = df["Humedad (%)"].max()
humedad_minima = df["Humedad (%)"].min()

# Humedad del suelo

humedad_suelo_promedio = df["Humedad Suelo (%)"].mean()

# Eventos de lluvia

eventos_lluvia = df["Lluvia"].sum()

print("=== INDICADORES DEL CULTIVO ===")
print(f"Temperatura promedio: {temperatura_promedio:.2f} °C")
print(f"Temperatura máxima: {temperatura_maxima:.2f} °C")
print(f"Temperatura mínima: {temperatura_minima:.2f} °C")

print(f"Humedad promedio: {humedad_promedio:.2f} %")
print(f"Humedad máxima: {humedad_maxima:.2f} %")
print(f"Humedad mínima: {humedad_minima:.2f} %")

print(f"Humedad promedio del suelo: {humedad_suelo_promedio:.2f} %")

print(f"Eventos de lluvia registrados: {eventos_lluvia}")
```

```python
def estado_suelo(h):
    if h < 65:
        return "Necesita riego"
    elif h <= 80:
        return "Adecuado"
    else:
        return "Exceso de humedad"
```

```python
def diagnostico_suelo(humedad):

    if humedad < 65:
        return "Necesita riego"

    elif humedad <= 80:
        return "Condicion adecuada"

    else:
        return "Exceso de humedad"


df["Diagnostico"] = df["Humedad Suelo (%)"].apply(diagnostico_suelo)

print(df[["Fecha",
          "Momento",
          "Humedad Suelo (%)",
          "Diagnostico"]])
```
